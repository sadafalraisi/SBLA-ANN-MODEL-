```python
from tensorflow.keras.layers import Input, Bidirectional, GRU, GlobalAveragePooling1D, Dense, Dropout
from tensorflow.keras.models import Model

from tensorflow.keras.layers import Input, LSTM, Bidirectional, Attention
from tensorflow.keras.layers import Dense, BatchNormalization, Flatten
from tensorflow.keras.models import Model
from tensorflow.keras.regularizers import l2

def create_model(input_shape, lstm_units, l2_reg,
                         dense1, dense2, n_classes):

    inputs = Input(shape=input_shape)

    x = Bidirectional(
        LSTM(
            lstm_units,
            return_sequences=True,
            kernel_regularizer=l2(l2_reg)
        )
    )(inputs)

    x = Attention()([x, x])

    x = Flatten()(x)

    x = Dense(dense1, activation="relu")(x)
    x = BatchNormalization()(x)

    x = Dense(dense2, activation="relu")(x)

    outputs = Dense(1, activation="sigmoid")(x)

    model = Model(inputs, outputs)

    return model


def model_complexity(model, input_shape):
    """
    Computes model complexity metrics.

    Parameters
    ----------
    model : tf.keras.Model
        Trained or untrained Keras model.
    input_shape : tuple
        Shape of one input sample (excluding batch size).
        Example: (122,), (32, 32, 3)

    Returns
    -------
    dict
        Total Parameters
        Trainable Parameters
        Non-trainable Parameters
        Model Size (MB)
        FLOPs (GFLOPs)
    """

    # Build model if not already built
    model.build((None, *input_shape))

    # -----------------------------
    # Parameters
    # -----------------------------
    total_params = model.count_params()

    trainable_params = np.sum(
        [tf.keras.backend.count_params(w) for w in model.trainable_weights]
    )

    non_trainable_params = np.sum(
        [tf.keras.backend.count_params(w) for w in model.non_trainable_weights]
    )

    # -----------------------------
    # Model Size
    # -----------------------------
    model_size_bytes = total_params * 4       # float32 = 4 bytes
    model_size_mb = model_size_bytes / (1024 ** 2)

    # -----------------------------
    # FLOPs
    # -----------------------------
    @tf.function
    def forward_pass(x):
        return model(x)

    concrete_func = forward_pass.get_concrete_function(
        tf.TensorSpec([1, *input_shape], tf.float32)
    )

    from tensorflow.python.framework.convert_to_constants import (
        convert_variables_to_constants_v2,
    )

    frozen_func = convert_variables_to_constants_v2(concrete_func)

    graph_def = frozen_func.graph.as_graph_def()

    with tf.Graph().as_default() as graph:
        tf.graph_util.import_graph_def(graph_def, name="")

        run_meta = tf.compat.v1.RunMetadata()

        opts = tf.compat.v1.profiler.ProfileOptionBuilder.float_operation()

        flops = tf.compat.v1.profiler.profile(
            graph=graph,
            run_meta=run_meta,
            cmd='op',
            options=opts
        )

    total_flops = flops.total_float_ops
    gflops = total_flops / 1e9

    return {
        "Total Parameters": int(total_params),
        "Trainable Parameters": int(trainable_params),
        "Non-trainable Parameters": int(non_trainable_params),
        "Model Size (MB)": round(model_size_mb, 3),
        "FLOPs": int(total_flops),
        "GFLOPs": round(gflops, 4)
    }
```

```python
from sklearn.metrics import (
    confusion_matrix,
    roc_curve,
    roc_auc_score
)
from sklearn.preprocessing import label_binarize
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


def plot_confusion_matrix(
    y_true,
    y_pred,
    class_names,
    title,
    save_path=None,
    normalize=False
):

    cm = confusion_matrix(y_true, y_pred)

    if normalize:
        cm = cm.astype(float) / (
            cm.sum(axis=1)[:, np.newaxis] + 1e-10
        )

    plt.figure(figsize=(8,6))

    sns.heatmap(
        cm,
        annot=True,
        fmt=".2f" if normalize else "d",
        cmap="Blues",
        xticklabels=class_names,
        yticklabels=class_names
    )

    plt.title(title)
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.tight_layout()

    if save_path:
        plt.savefig(save_path, dpi=700)

    plt.show()
    plt.close()



def plot_multiclass_roc(
    y_true,
    y_prob,
    class_names,
    title,
    save_path=None
):

    y_true = np.asarray(y_true).ravel()
    y_prob = np.asarray(y_prob)

    plt.figure(figsize=(7, 6))

    # --------------------------------------------------
    # Binary classification
    # --------------------------------------------------

    y_score = y_prob.ravel()

    # ROC curve
    fpr, tpr, _ = roc_curve(
        y_true,
        y_score
    )

    # ROC curve
    fpr, tpr, _ = roc_curve(
        y_true,
        y_score
    )

    auc_score = roc_auc_score(
        y_true,
        y_score
    )

    plt.plot(
        fpr,
        tpr,
        label=f"ROC (AUC = {auc_score:.4f})"
    )

    
    # --------------------------------------------------
    # Random classifier
    # --------------------------------------------------

    plt.plot(
        [0, 1],
        [0, 1],
        linestyle="--",
        label="Random"
    )

    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")

    plt.title(title)

    plt.legend()
    plt.grid(True)

    if save_path is not None:
        plt.savefig(
            save_path,
            dpi=300,
            bbox_inches="tight"
        )

    plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import (
    classification_report,
    accuracy_score,
    roc_auc_score,
    matthews_corrcoef,
    confusion_matrix
)
from sklearn.preprocessing import label_binarize

def classwise_metrics(y_true, y_pred, y_prob, class_names):

    y_true = np.asarray(y_true).ravel()
    y_pred = np.asarray(y_pred).ravel()
    y_prob = np.asarray(y_prob)

    classes = np.array([0, 1])

    results = []

    for cls in classes:

        # One-vs-rest labels
        yt = (y_true == cls).astype(int)
        yp = (y_pred == cls).astype(int)

        y_score = y_prob.ravel()
        
        # Confusion matrix
        tn, fp, fn, tp = confusion_matrix(
            yt,
            yp,
            labels=[0, 1]
        ).ravel()

        # Precision
        precision = (
            tp / (tp + fp)
            if (tp + fp) > 0
            else 0
        )

        # Recall / sensitivity
        recall = (
            tp / (tp + fn)
            if (tp + fn) > 0
            else 0
        )

        # Specificity
        specificity = (
            tn / (tn + fp)
            if (tn + fp) > 0
            else 0
        )

        # F1
        f1 = (
            2 * precision * recall / (precision + recall)
            if (precision + recall) > 0
            else 0
        )

        # FPR
        fpr = (
            fp / (fp + tn)
            if (fp + tn) > 0
            else 0
        )

        # G-Mean
        gmean = np.sqrt(
            recall * specificity
        )

        # MCC
        denominator = np.sqrt(
            (tp + fp) *
            (tp + fn) *
            (tn + fp) *
            (tn + fn)
        )

        mcc = (
            ((tp * tn) - (fp * fn)) / denominator
            if denominator > 0
            else 0
        )

        # ROC-AUC
        if len(np.unique(yt)) == 2:
            roc_auc = roc_auc_score(
                yt,
                y_score
            )
        else:
            roc_auc = np.nan

        # Support
        support = np.sum(yt)

        results.append({
            "Class": class_names[cls],
            "Precision": precision,
            "Recall": recall,
            "F1": f1,
            "FPR": fpr,
            "GMean": gmean,
            "MCC": mcc,
            "ROC_AUC": roc_auc,
            "Support": support
        })

    return pd.DataFrame(results)
def calculate_metrics(
    y_true,
    y_pred,
    y_prob
):

    # Calculate multiclass FPR using One-vs-Rest
    classes = np.unique(y_true)

    fprs = []
    supports = []

    for cls in classes:

        # One-vs-Rest transformation
        yt = (np.array(y_true) == cls).astype(int)
        yp = (np.array(y_pred) == cls).astype(int)

        tn, fp, fn, tp = confusion_matrix(yt, yp).ravel()

        # False Positive Rate
        fpr = fp / (fp + tn) if (fp + tn) > 0 else 0

        fprs.append(fpr)
        supports.append(np.sum(yt))

    # Weighted FPR
    weighted_fpr = np.average(
        fprs,
        weights=supports
    )

    metrics = {

        "Accuracy":
            accuracy_score(
                y_true,
                y_pred
            ),
    
        "Precision":
            precision_score(
                y_true,
                y_pred,
                average="weighted",
                zero_division=0
            ),
    
        "Recall":
            recall_score(
                y_true,
                y_pred,
                average="weighted",
                zero_division=0
            ),
    
        "F1":
            f1_score(
                y_true,
                y_pred,
                average="weighted",
                zero_division=0
            ),
    
        "FPR":
            weighted_fpr,
    
        "ROC_AUC":
            roc_auc_score(
                np.asarray(y_true).ravel(),
                np.asarray(y_prob).ravel()
            ),
    
        "GMean":
            geometric_mean_score(
                y_true,
                y_pred
            ),
    
        "MCC":
            matthews_corrcoef(
                y_true,
                y_pred
            )
    }
    return pd.DataFrame([metrics])
def timing_report(
    train_time,
    test_time,
    predict_time
):

    return pd.DataFrame({

        "Training Time (s)":
            [train_time],

        "Testing Time (s)":
            [test_time],

        "Prediction Time (s)":
            [predict_time],

        "Total Time (s)":
            [
                train_time +
                test_time +
                predict_time
            ]

    })

```

```python
def evaluate_dataset(
    model,
    X,
    y,
    dataset_name,
    class_names
):

    print(f"\n{'='*60}")
    print(dataset_name.upper())
    print(f"{'='*60}")

    start = time.time()

    y_prob = model.predict(X, verbose=0)

    # Binary classification
    # y_prob shape = (n_samples, 2)
    y_score = y_prob.ravel()

    # Convert probability of class 1 to binary prediction
    y_pred = (y_score >= 0.5).astype(int)

    pred_time = time.time() - start

    metrics_df = calculate_metrics(
        y,
        y_pred,
        y_prob
    )

    display(metrics_df.round(4))

    class_df = classwise_metrics(
        y,
        y_pred,
        y_prob,
        class_names
    )

    print("\nClass-wise Performance")
    display(class_df)

    plot_confusion_matrix(
        y,
        y_pred,
        class_names,
        f"{dataset_name} Confusion Matrix"
    )

    plot_confusion_matrix(
        y,
        y_pred,
        class_names,
        f"{dataset_name} Normalized CM",
        normalize=True
    )

    plot_multiclass_roc(
        y,
        y_prob,
        class_names,
        f"{dataset_name} ROC Curve"
    )

    return metrics_df, class_df, pred_time
```

```python
import subprocess


def get_gpu_usage():

    try:
        result = subprocess.check_output(
            [
                "nvidia-smi",
                "--query-gpu=utilization.gpu",
                "--format=csv,noheader,nounits"
            ],
            stderr=subprocess.DEVNULL
        ).decode("utf-8").strip()

        if not result:
            return 0.0

        # If multiple GPUs exist, return average utilization
        values = [
            float(x.strip())
            for x in result.splitlines()
            if x.strip()
        ]

        return sum(values) / len(values)

    except (subprocess.CalledProcessError, FileNotFoundError, ValueError):
        return 0.0
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import roc_curve, auc


# =========================================================
# AGGREGATE FOLD METRICS
# =========================================================

def aggregate_fold_metrics(fold_results):

    rows = []

    for fold_idx, result in enumerate(fold_results, start=1):

        for dataset_name, key in [
            ("Train", "train_metrics"),
            ("Validation", "val_metrics"),
            ("Test", "test_metrics")
        ]:

            metrics_df = result[key].copy()

            row = {
                "Fold": fold_idx,
                "Dataset": dataset_name
            }

            for col in metrics_df.columns:

                if col != "Dataset":

                    value = metrics_df[col].iloc[0]

                    try:
                        row[col] = float(value)
                    except:
                        pass

            rows.append(row)

    fold_metric_df = pd.DataFrame(rows)

    metric_columns = [
        col
        for col in fold_metric_df.columns
        if col not in ["Fold", "Dataset"]
    ]

    mean_df = (
        fold_metric_df
        .groupby("Dataset")[metric_columns]
        .mean()
    )

    std_df = (
        fold_metric_df
        .groupby("Dataset")[metric_columns]
        .std()
    )

    summary_df = mean_df.copy()

    for col in metric_columns:

        summary_df[col] = (
            mean_df[col].round(4).astype(str)
            + " ± "
            + std_df[col].round(4).astype(str)
        )

    return fold_metric_df, mean_df, std_df, summary_df


# =========================================================
# AGGREGATE TIME
# =========================================================

def aggregate_time(fold_results):

    rows = []

    for fold_idx, result in enumerate(fold_results, start=1):

        rows.append({
            "Fold": fold_idx,
            "Train Time": result["train_time"],
            "Validation Evaluation Time": result["val_time"],
            "Test Evaluation Time": result["test_time"],
            "Train Prediction Time": result["train_pred_time"],
            "Validation Prediction Time": result["val_pred_time"],
            "Test Prediction Time": result["test_pred_time"],
            "Total Time": result["total_time"]
        })

    df = pd.DataFrame(rows)

    mean = df.drop(columns="Fold").mean()
    std = df.drop(columns="Fold").std()

    summary = pd.DataFrame({
        "Mean": mean,
        "Std": std
    })

    return df, summary


# =========================================================
# AGGREGATE MEMORY
# =========================================================

def aggregate_memory(fold_results):

    rows = []

    for fold_idx, result in enumerate(fold_results, start=1):

        rows.append({
            "Fold": fold_idx,
            "After Model": result["mem_after_model"],
            "After Train": result["mem_after_train"],
            "After Test": result["mem_after_test"],
            "After Validation": result["mem_after_val"],
            "After Prediction": result["mem_end"]
        })

    df = pd.DataFrame(rows)

    mean = df.drop(columns="Fold").mean()
    std = df.drop(columns="Fold").std()

    summary = pd.DataFrame({
        "Mean_MB": mean,
        "Std_MB": std
    })

    return df, summary


# =========================================================
# AGGREGATE CONFUSION MATRICES
# =========================================================

def aggregate_confusion_matrices(fold_results):

    train_cm = np.mean(
        [r["train_cm"] for r in fold_results],
        axis=0
    ).round().astype(int)

    val_cm = np.mean(
        [r["val_cm"] for r in fold_results],
        axis=0
    ).round().astype(int)

    test_cm = np.mean(
        [r["test_cm"] for r in fold_results],
        axis=0
    ).round().astype(int)

    return train_cm, val_cm, test_cm


# =========================================================
# PLOT CONFUSION MATRIX
# =========================================================

def plot_cm(cm, title, normalize=False):

    if normalize:
        cm = cm.astype(float) / (
            cm.sum(axis=1)[:, np.newaxis] + 1e-10
        )

    plt.figure(figsize=(8, 6))

    sns.heatmap(
        cm,
        annot=True,
        fmt=".2f" if normalize else "d",
        cmap="Blues",
        xticklabels=CLASS_NAMES,
        yticklabels=CLASS_NAMES
    )

    plt.title(title)
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.tight_layout()

    plt.show()
    plt.close()


# =========================================================
# BINARY ROC
# =========================================================

def plot_aggregated_roc(
    fold_results,
    dataset="test",
    model_name="BSLA-ANN",
    class_names=CLASS_NAMES
):

    y_true_all = []
    y_prob_all = []

    for result in fold_results:

        y_true_all.append(
            result[f"{dataset}_y_true"]
        )

        probabilities = result[
            f"{dataset}_probabilities"
        ]

        y_prob_all.append(
            probabilities.ravel()
        )

    y_true_all = np.concatenate(y_true_all)
    y_prob_all = np.concatenate(y_prob_all)

    fpr, tpr, thresholds = roc_curve(
        y_true_all,
        y_prob_all
    )

    roc_auc = auc(fpr, tpr)

    plt.figure(figsize=(8, 6))

    plt.plot(
        fpr,
        tpr,
        label=f"5-Fold ROC (AUC = {roc_auc:.4f})"
    )

    plt.plot(
        [0, 1],
        [0, 1],
        linestyle="--"
    )

    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")

    plt.title(
        plt.title(f"5-Fold ROC - {model_name} - {dataset.capitalize()}")
    )

    plt.legend()
    plt.grid()
    plt.tight_layout()

    plt.show()
    plt.close()

    return roc_auc


# =========================================================
# AGGREGATE GPU
# =========================================================

def aggregate_gpu(fold_results):

    rows = []

    for fold_idx, result in enumerate(fold_results, start=1):

        rows.append({
            "Fold": fold_idx,
            "Before Model": result["gpu_before_model"],
            "After Model": result["gpu_after_model"],
            "After Train": result["gpu_after_train"],
            "After Test": result["gpu_after_test"],
            "After Validation": result["gpu_after_val"],
            "After Prediction": result["gpu_end"]
        })

    df = pd.DataFrame(rows)

    mean = df.drop(columns="Fold").mean()
    std = df.drop(columns="Fold").std()

    summary = pd.DataFrame({
        "Mean_GPU": mean,
        "Std_GPU": std
    })

    return df, summary
```
